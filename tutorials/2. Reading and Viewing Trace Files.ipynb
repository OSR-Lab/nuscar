# 📑Tutorial-2: Reading and Viewing Side-Channel Traces

**📈 traceset trace management module**

Side-channel analysis is an attack technique that exploits side-channel information leakage (such as power consumption and electromagnetic emanations) produced while a cryptographic device performs computations, in order to analyze cryptographic algorithms. Side-channel leakage is generally measured with an oscilloscope; we refer to this as the signal samples (Samples), while the input/output and other information required for the algorithm's computation, such as plaintext, ciphertext, and IV, is referred to as metadata (Metadata). A side-channel trace consists of both sample data and metadata.

**Side-channel trace = Signal samples (Samples) + Metadata (Metadata)**

In side-channel analysis, multiple side-channel traces are generally required for statistical analysis, and we store them in files.

The default trace format in NUSCAR is a custom Zarr format, managed with [Zarr-Python](https://zarr.readthedocs.io/en/stable/), and it is also compatible with the [ETS format](https://github.com/eshard/estraces).

###  Defining the reader and container

```python
# Import the required libraries
import nuscar
import numpy as np
import time
```

```python
reader = nuscar.ReaderZARR("../datasets/dpa_v2.zarr") # Create the reader 
ctn = nuscar.ContainerZARR(reader)                    # Create the container; when defining the container you can specify the sample range (frame) to read and preprocessing methods (such as filtering)
```

```python
ctn  # The container object displays basic information in a notebook environment
```

```python
ctn[0].samples
```

### Viewing Metadata information

```python
ctn.viewmeta([1,2,5]) # View the meta information of traces 1, 2, and 5
```

### Trace visualization

```python
fig = ctn.view([1,2,5], resample=False, webgl=False) # Display the sample data of traces 1, 2, and 5
fig
```

### Accessing trace data

```python
p = ctn[10:12].plaintext 
s = ctn[[1,2,10]].samples # With the NUSCAR API, you can conveniently access traces and metadata at different indices
print(p)
print(s)
```
