# 📑Tutorial-1: Interactive Visualization

**📈 plot: Interactive Visualization Module**

Inspecting traces is essential in side-channel analysis. Analysts need an interactive visualization plotting system to intuitively understand side-channel traces, perform SPA, and locate the position of cryptographic operations. NUSCAR's Plotly-based intelligent plotting engine enables interactive visualization of large amounts of sampled data, greatly improving the efficiency of side-channel analysis.


### Basic Plotting

```python
import nuscar
import numpy as np
```

```python
x = np.linspace(0, 20 * np.pi, 1000)
y1 = np.sin(x)
y2 = np.cos(x)
y = np.vstack([y1, y2])
nuscar.plot(x=x, y=y, name=['sin(x)', 'cos(x)'], title='sin & cos', resample=False, webgl=False, shift=False)
```

Use the plot function to draw traces, where:
- the resample parameter controls whether the object is automatically resampled, default is automatic;
- the webgl parameter controls whether WebGL is used to accelerate rendering, default is yes;
- the shift parameter controls whether the slider is shown, which can be used for manually dragging traces into alignment, default is not shown;

Below the figure, the rangeslider can show a global overview. Due to plotly limitations, the rangeslider is unavailable when resample or webgl is enabled.

By default the mouse can drag and zoom along the x-axis; holding the `y` key enables dragging and zooming along the y-axis.

### Plotting Large Amounts of Data 

```python
bigy = np.random.randn(2, 5000_000)
fig = nuscar.plot(bigy, resample=True, webgl=True) # Using resample and webgl enables smooth visualization of large amounts of data
fig 
```

### Manual Drag Alignment

```python
fig = nuscar.plot(x=x, y=y, name=['sin(x)', 'cos(x)'], title='sin & cos', resample=False, webgl=False, shift=True)
fig 
```

### Plotting Pattern and Peak Positions

In side-channel analysis, it is often necessary to locate a pattern or some peak points through observation. In NUSCAR, we provide some helper functions to visualize this information.

```python
x = range(100)
y = np.sin(x)

fig = nuscar.plot_pattern(y, pattern_x=range(10, 20), resample=False, webgl=False) # Only one-dimensional data input is supported here
fig
```

```python
peaks = [42, 80, 83] # The x-coordinates of the peak positions of interest
```

```python
fig = nuscar.plot_peak(y, peak_x=peaks, resample=False, webgl=False) # Only one-dimensional data input is supported here
fig
```

### Plotting the Spectrogram

```python
x = range(1000)
y = np.sin(x)
fig = nuscar.plot_spectrogram(y)  # fs can adjust the sampling rate, fslim restricts the frequency
fig
```
